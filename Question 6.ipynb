```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
```

```python
x=np.linspace(-5,5,100)
y=np.linspace(-5,5,100)
x, y = np.meshgrid(x,y) #creates a grid of  x and y vales
z= np.sin(x**2+y**2) #compute z value as the sine of the sum of squares of x and y
```

```python
fig=plt.figure(figsize=(10,6))
ax=plt.axes(projection='3d') #creates a 3D axes using projection='3d' with standard axes creation method
surface=ax.plot_surface(x,y,z,cmap='coolwarm') #creates a 3D surface plot
fig.colorbar(surface, shrink=0.5) #adds a color bar to the plot

ax.set_xlabel('$x$') #sets label for x axes
ax.set_ylabel('$y$') #sets label for y axes
ax.set_zlabel('$z$') #sets label for z axes

plt.show()
```
