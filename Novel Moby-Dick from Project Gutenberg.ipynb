```python
import matplotlib.pyplot as plt
from collections import Counter #counts hashable objects
import requests #python library used for making HTTP requests
import string 
```

```python
url = 'https://www.gutenberg.org/files/2701/2701-0.txt' #the URL of the text of the novel Moby-Dick from Project Gutenberg.
content = requests.get(url) #sends an HTTP GET request to download the file at the specified URL.
text = content.text #extracts the text of Moby-Dick from the HTTP response message and saves it as a content. This is stored as a string in the variable text.
```

```python
text=text.upper() #converts all the text into uppercase
```

```python
alphabets='ABCDEFGHIJKLMNOPQRSTUVWXYZ' 
count_dict = {} #creates an empty dictionary
```

```python
for char in alphabets:
    count_dict[char] = text.count(char) #iterate through the alphabets and count the occurrence of each letter in the text
```

```python
total_letters = sum(count_dict.values()) #counts the total number of letters
```

```python
letters = sorted(count_dict.keys()) #sorts the alphabets in ascending order(A to Z)
percentages = [(count_dict[i] / total_letters) * 100 for i in letters] #calculates the percentage frequencies of each alphabet
```

```python
plt.figure(figsize=(10, 6))

plt.bar(letters, percentages, color='b') #creates a bar chart

plt.xlabel('Letters') #set the label for x-axis
plt.ylabel('Percentage (%)') #set the label for y-axis
plt.title('Letter Frequencies in Moby-Dick (Percentage)') #set the label for z-axis

plt.grid(axis='y', linestyle='--',color='gray') #creates a grid for y axis with dashed gray lines

plt.show() #displays the plot
```
