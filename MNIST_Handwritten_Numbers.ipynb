```python
import matplotlib.pyplot as plt
from sklearn.datasets import fetch_openml
```

```python
X, y= fetch_openml('mnist_784', return_X_y=True, as_frame=False)
```

```python
images = X[:50].reshape(-1, 28, 28) #selects the first 50 images from the dataset and reshapes each image data in the flattened form into a 28 x 28 2D array
labels = y[:50] #extracts the corresponding labels for the first 50 images.
plt.figure(figsize=(10, 5)) #size of the entire grid with images
```

```python
for i in range(50): #loops through the first 50 images
    axes = plt.subplot(5, 10, i + 1) #Creates a subplot for each image in a grid with 5 rows and 10 columns.
    plt.imshow(images[i], cmap='gray') #displays each image using imshow() function
    plt.axis('off') #hides the axes for each subplot
    plt.text(13.5, 32 , labels[i], fontsize=7, ha='center') #adds the corresponding digit label below the image. pass x and y coordinates in the text() function to place the label. 
plt.show()
```
