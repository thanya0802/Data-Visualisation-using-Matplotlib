```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
X,y =load_iris(return_X_y=True)
```

```python
sepal_length= X[:,0] #extracts the first column from X, corresponding to each flower's sepal length.
sepal_width= X[:,1]  #extracts the second column from X, corresponding to each flower's sepal width.
petal_width= X[:,3]  #extracts the fourth column from X, corresponding to each flower's petal width.
```

```python
plt.figure()

plt.scatter(sepal_length, sepal_width, s = petal_width*100, c=y, alpha=0.4) #creates a scatter plot

plt.xlabel('Sepal length(cm)')
plt.ylabel('Sepal width(cm)')
plt.title('Iris flower data set')

plt.grid() #creates a grid

plt.colorbar() #adds a colorbar
plt.show()
```
