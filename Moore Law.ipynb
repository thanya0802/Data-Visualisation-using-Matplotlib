```python
import numpy as np
import matplotlib.pyplot as plt
from math import log #imports log from math module
```

```python
years=np.arange(1972, 2014, 2) #creates a numpy array
```

```python
observed = np.log10(np.array([0.0025, 0.005, 0.029, 0.12, 0.275, 1.18, 3.1, 7.5, 24.0, 42.0, 220.0, 592.0, 1720.0, 2046.0, 3100.0])*1000000) #multiplies each observed transistor count by 1 million to convert from millions to absolute numbers and computes base 10 logarithm
observed_years = np.array([1972, 1974, 1978, 1982, 1985, 1989, 1993, 1997, 1999, 2000, 2003, 2004, 2007, 2008, 2012])
```

```python
n0 = 2500 #sets the initial transistor count in 1972 to 2500
y0 = years[0] #accesses the initial year and stores it.i.e. 1972
```

```python
predicted = log(n0, 10)  +  ((years - y0)/2)*log(2,10) #formula to predict transistor count values using Moore's law
```

```python
plt.figure(figsize=(10, 6))
```

```python
plt.plot(observed_years, observed, 'r*', label='Observed', markersize=20) #plots the observed data using red stars
plt.plot(years, predicted, 'k--', label='Predicted', markersize=50) #plots the predicted data using black dashed lines

plt.xlabel('Year') #adds the x label to the plot
plt.ylabel('log(n_trans)') #adds the y label to the plot
plt.title("Moore's Law") #sets the title of the plot
plt.legend() #adds legend to the plot
plt.show() #displays the plot
```
