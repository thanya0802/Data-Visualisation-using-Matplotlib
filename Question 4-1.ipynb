```python
import matplotlib.pyplot as plt
```

```python
labels = ['Chrome', 'Safari', 'Edge', 'Firefox','Samsung Internet', 'Opera']
x = [65.18, 18.55, 5.26, 2.74, 2.56, 2.15]
explode = (0.1, 0, 0, 0, 0, 0) #use explode parameter to seperate the "Chrome" slice slightly from the rest of the pie. 0.1 value will move the slice outwards by 10% of the radius.
plt.pie(x, labels= labels, autopct= '%.2f%%', explode = explode) #creates a pie chart
plt.title('Market Share of each browser worldwide in 2024')
```
